```python
import matplotlib.pyplot as plt
import numpy as np

# Data from the observation
labels = [
    "Binary with Conditional Distance",  # 53.8%
    "Binary Encoding",                   # 32.5%
    "Categorical Encoding",              # 7.6%
    "Mutually Exclusive Counts"          # 6.1%
]
sizes = [106, 64, 15, 12]

# Custom Color Scheme based on the "Salford & Co." Theme
# - Primary Gold/Ochre (matches "DATA" text)
# - Muted Beige/Taupe (Secondary)
# - Dark Brown (Accent)
# - Light Cream (Highlight)
colors = ['#CCA458', '#D8C6A8', '#8C7456', '#F0EAD6']

# Background color
bg_color = '#efe1ce'
# Text color (Dark Brown for readability)
text_color = '#594A38'

# Create the plot
fig, ax = plt.subplots(figsize=(10, 7), facecolor=bg_color)

# Create pie chart with percentages only
wedges, texts, autotexts = ax.pie(
    sizes,
    colors=colors,
    startangle=90,
    counterclock=False,
    autopct='%1.1f%%',
    textprops={'color': '#2d2d2d', 'fontsize': 12, 'weight': 'bold'}
)

# Ensure the pie chart is circular
plt.axis('equal')

# Save the plot
plt.savefig('themed_pie_chart.png', bbox_inches='tight', facecolor=bg_color)
plt.show()
```
