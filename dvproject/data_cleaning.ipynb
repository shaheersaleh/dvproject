## 4. Section 2: Agriculture & Irrigation Columns

**Cleaning Strategy:** Binary encoding (1 = present, 0 = absent) + Descriptive column renaming

**Crop Columns:** Wheat, Rice, Cotton, Sugarcane, Maize, Pulses, Orchard, Vegetables  
**Irrigation Columns:** Canal, Rive, Tubwell/Well, Ravine/Stream, Spring/Karez, Arid, Flooding Torrent, Others

**Issue:** Numeric values indicate presence but numbers themselves have no significance  
**Action:** 
1. Convert any non-empty value to 1, empty/NaN to 0
2. Rename crop columns to "Major Crop [Name]"
3. Rename irrigation columns to "Irrigation System [Name]"

```python
# Inspect before cleaning
agri_columns = ['Wheat', 'Rice', 'Cotton', 'Sugarcane', 'Maize', 'Pulses', 'Orchard', 
                'Vegetables', 'Canal', 'Rive', 'Tubwell/Well', 'Ravine/Stream', 
                'Spring/Karez', 'Arid ', 'Flooding Torrent', 'Others']

print("BEFORE CLEANING - Agriculture & Irrigation Columns:")
for col in agri_columns:
    print(f"{col}: {df[col].isnull().sum()} missing, {df[col].nunique()} unique values")
print(f"\nSample values from Wheat column:")
print(df['Wheat'].value_counts())
```

```python
# Step 1: Convert to binary encoding
# Rename 'Rive' to 'River' first
if 'Rive' in df.columns:
    df.rename(columns={'Rive': 'River'}, inplace=True)
if 'Rive' in agri_columns:
    agri_columns[agri_columns.index('Rive')] = 'River'

# STEP 1: Track rows where ALL agriculture columns are empty (group-based NaN logic)
mask_all_empty = df[agri_columns].isnull().all(axis=1) | \
                 ((df[agri_columns] == 0).all(axis=1)) | \
                 ((df[agri_columns] == '0').all(axis=1)) | \
                 ((df[agri_columns] == '').all(axis=1))

# STEP 2: Convert each column to binary (1 if has value, 0 if empty/NaN)
for col in agri_columns:
    df[col] = df[col].replace('', np.nan)
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].notna().astype(int)

# STEP 3: Rename columns with descriptive prefixes
crop_columns = {
    'Wheat': 'Major Crop Wheat',
    'Rice': 'Major Crop Rice',
    'Cotton': 'Major Crop Cotton',
    'Sugarcane': 'Major Crop Sugarcane',
    'Maize': 'Major Crop Maize',
    'Pulses': 'Major Crop Pulses',
    'Orchard': 'Major Crop Orchard',
    'Vegetables': 'Major Crop Vegetables'
}

irrigation_columns = {
    'Canal': 'Irrigation System Canal',
    'River': 'Irrigation System River',
    'Tubwell/Well': 'Irrigation System Tubwell/Well',
    'Ravine/Stream': 'Irrigation System Ravine/Stream',
    'Spring/Karez': 'Irrigation System Spring/Karez',
    'Arid ': 'Irrigation System Arid',
    'Flooding Torrent': 'Irrigation System Flooding Torrent',
    'Others': 'Irrigation System Others'
}

# Apply renaming
df.rename(columns={**crop_columns, **irrigation_columns}, inplace=True)

# STEP 4: Restore NaN for rows where ALL agriculture columns were empty
agri_columns_new = list(crop_columns.values()) + list(irrigation_columns.values())
for col in agri_columns_new:
    df.loc[mask_all_empty, col] = np.nan

print("AFTER CLEANING & RENAMING - Agriculture & Irrigation Columns:")
for col in agri_columns_new:
    present = df[col].sum()
    absent = (df[col] == 0).sum()
    nan = df[col].isnull().sum()
    print(f"{col}: Present={present} (1's), Absent={absent} (0's), NaN={nan}")
```

### Section 4: Modern Technology for Irrigation

**Cleaning Strategy:** Conditional numeric cleaning based on "None" indicator

**Columns:**
1. Use of Modren Tech. for 1-Irrigation Drip System No → Number of Drip Systems
2. Sprinkler System No → Number of Sprinkler Systems
3. Others (Specify) No → Number of Other Systems
4. None → Indicator (4 = none used)

**Action:**
1. If "None" = 4, convert to 1 (indicator that no systems used), set all system columns to 0
2. If "None" is empty/NaN, keep all system columns as NaN
3. Otherwise, convert system columns to numeric (empty → NaN, keep numbers)
4. Rename columns appropriately

```python
# Inspect before cleaning
drip_col = 'Use of Modren Tech. for 1-Irrigation Drip System No'
sprinkler_col = 'Sprinkler System No'
others_col = 'Others (Specify) No'
none_col = 'None'

print("BEFORE CLEANING - Modern Technology for Irrigation:")
print(f"\n1. {drip_col}")
print(f"   Missing: {df[drip_col].isnull().sum()}")
print(f"   Value counts:\n{df[drip_col].value_counts().head()}")

print(f"\n2. {sprinkler_col}")
print(f"   Missing: {df[sprinkler_col].isnull().sum()}")
print(f"   Value counts:\n{df[sprinkler_col].value_counts().head()}")

print(f"\n3. {others_col}")
print(f"   Missing: {df[others_col].isnull().sum()}")
print(f"   Value counts:\n{df[others_col].value_counts().head()}")

print(f"\n4. {none_col}")
print(f"   Missing: {df[none_col].isnull().sum()}")
print(f"   Value counts:\n{df[none_col].value_counts().head()}")
```

```python
# Clean: Modern Technology columns
drip_col = 'Use of Modren Tech. for 1-Irrigation Drip System No'
sprinkler_col = 'Sprinkler System No'
others_col = 'Others (Specify) No'
none_col = 'None'

# Step 1: Clean the "None" indicator column
df[none_col] = df[none_col].replace('', np.nan)
df[none_col] = pd.to_numeric(df[none_col], errors='coerce')

# Step 2: Clean the system count columns
for col in [drip_col, sprinkler_col, others_col]:
    df[col] = df[col].replace('', np.nan)
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Step 3: Apply conditional logic based on "None" column
# If None = 4 (no systems used), set all system columns to 0 and None to 1
mask_none = df[none_col] == 4
df.loc[mask_none, drip_col] = 0
df.loc[mask_none, sprinkler_col] = 0
df.loc[mask_none, others_col] = 0
df.loc[mask_none, none_col] = 1

# If None is NaN (empty), keep all system columns as NaN
mask_none_nan = df[none_col].isna()
df.loc[mask_none_nan, drip_col] = np.nan
df.loc[mask_none_nan, sprinkler_col] = np.nan
df.loc[mask_none_nan, others_col] = np.nan

# Step 4: Rename columns
df.rename(columns={
    drip_col: 'Number of Drip Systems',
    sprinkler_col: 'Number of Sprinkler Systems',
    others_col: 'Number of Other Irrigation Systems',
    none_col: 'No Modern Irrigation System'
}, inplace=True)

print("AFTER CLEANING - Modern Technology for Irrigation:")
print(f"\n1. Number of Drip Systems:")
print(f"   Missing (NaN): {df['Number of Drip Systems'].isnull().sum()}")
print(f"   Zero values: {(df['Number of Drip Systems'] == 0).sum()}")
print(f"   Value counts (top 5):\n{df['Number of Drip Systems'].value_counts().head()}")

print(f"\n2. Number of Sprinkler Systems:")
print(f"   Missing (NaN): {df['Number of Sprinkler Systems'].isnull().sum()}")
print(f"   Zero values: {(df['Number of Sprinkler Systems'] == 0).sum()}")

print(f"\n3. Number of Other Irrigation Systems:")
print(f"   Missing (NaN): {df['Number of Other Irrigation Systems'].isnull().sum()}")
print(f"   Zero values: {(df['Number of Other Irrigation Systems'] == 0).sum()}")

print(f"\n4. No Modern Irrigation System:")
print(f"   Value counts:\n{df['No Modern Irrigation System'].value_counts()}")
```

### Section 5: Depth of Water Table in Mouza

**Cleaning Strategy:** Convert to numeric, preserve NaN for missing values, rename to include unit

**Column:** `Depth of Water Table in Mouza` → `Depth of Water Table in ft`

**Columns:** 14 playground types (7 male + 7 female)
- Male: Cricket, Hockey, Football, Volleyball, Kabbady, Other Sports, None
- Female: Cricket, Hockey, Football, Volleyball, Kabbady, Other Sports, None

**Logic:**
- Convert to binary: Any value present → 1 (Yes), empty → keep as 0 or NaN
- 0/empty → NaN when no data available
- Remove header rows (P4Q...)
- Special None column logic: If None=1 (has value), then all other sports in that gender should be 0
- NaN only when data not available (all columns empty for that gender)

### Section 14: Veterinary Health Care Facilities

**Cleaning Strategy:** Binary for availability + conditional distance (same as Health Facilities)

**Columns:**
1. Dispensary/Veterinary Centre (Yes + Distance)
2. Private Facility (Yes + Distance)  
3. Mobile Veterinary Dispenser (availability only)

**Logic:**
- Yes columns: 1=Yes, 2→0 (No), 0/empty→NaN, remove header rows
- Distance columns: Convert to numeric km, remove header rows
- If Yes=1: Distance=0 (facility available in mouza)
- If Yes=0: Keep distance value, fill missing with 0
- If Yes=NaN: Distance=NaN (no data)
- Mobile Dispenser: Binary only (no distance column)

```python
# Inspect Playground columns
male_playgrounds = [
    ('Regular Playgrounds in Mouza \n1-Male \n1-Cricket', 'Male Cricket Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n2-Hockey', 'Male Hockey Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n3-Foot Ball', 'Male Football Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n4-Volley Ball', 'Male Volleyball Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n5-Kabaddy', 'Male Kabbady Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n6-Any Other Sports', 'Male Other Sports Playground'),
    ('Regular Playgrounds in Mouza \n1-Male \n7-None', 'Male None Playground')
]

female_playgrounds = [
    ('Regular Playgrounds in Mouza \n1-Female \n1-Cricket', 'Female Cricket Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n2-Hockey', 'Female Hockey Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n3-Foot Ball', 'Female Football Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n4-Volley Ball', 'Female Volleyball Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n5-Kabaddy', 'Female Kabbady Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n6-Any Other Sports', 'Female Other Sports Playground'),
    ('Regular Playgrounds in Mouza \n1-Female \n7-None', 'Female None Playground')
]

all_playgrounds = male_playgrounds + female_playgrounds

print("BEFORE CLEANING - Playgrounds:")
print("=" * 80)
for col, name in all_playgrounds:
    print(f"\n{name}:")
    print(f"  Missing: {df[col].isnull().sum()}, Zeros: {(df[col] == 0).sum() + (df[col] == '0').sum()}")
    print(f"  Values: {df[col].value_counts().to_dict()}")
```

```python
# Clean Playground columns
# Step 1: Identify rows where ALL male playgrounds are empty
male_playground_cols = [col for col, name in male_playgrounds]
mask_male_all_empty = df[male_playground_cols].isnull().all(axis=1) | \
                       ((df[male_playground_cols] == 0).all(axis=1)) | \
                       ((df[male_playground_cols] == '0').all(axis=1)) | \
                       ((df[male_playground_cols] == '').all(axis=1))

# Step 2: Identify rows where ALL female playgrounds are empty
female_playground_cols = [col for col, name in female_playgrounds]
mask_female_all_empty = df[female_playground_cols].isnull().all(axis=1) | \
                         ((df[female_playground_cols] == 0).all(axis=1)) | \
                         ((df[female_playground_cols] == '0').all(axis=1)) | \
                         ((df[female_playground_cols] == '').all(axis=1))

# Step 3: Clean all columns - remove headers, convert to binary
for col, name in all_playgrounds:
    # Remove header row values (like 'P4Q0211')
    df[col] = df[col].apply(lambda x: np.nan if (isinstance(x, str) and x.startswith('P4Q')) else x)
    # Replace empty strings with NaN temporarily
    df[col] = df[col].replace('', np.nan)
    # Replace 0 with NaN temporarily
    df[col] = df[col].replace('0', np.nan)
    df[col] = df[col].replace(0, np.nan)
    # Convert any non-NaN value to 1 (Yes/Present)
    df[col] = df[col].notna().astype(float)
    df[col] = df[col].replace(0, np.nan)  # Convert 0 back to NaN (no data)

# Step 4: Handle "None" column logic for Male playgrounds
male_none_col = 'Regular Playgrounds in Mouza \n1-Male \n7-None'
male_sport_cols = [col for col, name in male_playgrounds if 'None' not in name]

# Where Male None = 1, set all other male sports to 0
mask_male_none = df[male_none_col] == 1
for col in male_sport_cols:
    df.loc[mask_male_none, col] = 0

# Where any male playground has data (not in all_empty group), fill missing with 0
mask_male_has_data = ~mask_male_all_empty
for col in male_playground_cols:
    df.loc[mask_male_has_data, col] = df.loc[mask_male_has_data, col].fillna(0)

# Step 5: Handle "None" column logic for Female playgrounds
female_none_col = 'Regular Playgrounds in Mouza \n1-Female \n7-None'
female_sport_cols = [col for col, name in female_playgrounds if 'None' not in name]

# Where Female None = 1, set all other female sports to 0
mask_female_none = df[female_none_col] == 1
for col in female_sport_cols:
    df.loc[mask_female_none, col] = 0

# Where any female playground has data (not in all_empty group), fill missing with 0
mask_female_has_data = ~mask_female_all_empty
for col in female_playground_cols:
    df.loc[mask_female_has_data, col] = df.loc[mask_female_has_data, col].fillna(0)

# Step 6: Restore NaN for rows where ALL playgrounds were originally empty
for col in male_playground_cols:
    df.loc[mask_male_all_empty, col] = np.nan

for col in female_playground_cols:
    df.loc[mask_female_all_empty, col] = np.nan

# Step 7: Rename columns
rename_dict = {col: name for col, name in all_playgrounds}
df.rename(columns=rename_dict, inplace=True)

print("AFTER CLEANING - Playgrounds:")
print("=" * 80)
for _, name in all_playgrounds:
    print(f"\n{name}:")
    print(f"  Values: {df[name].value_counts().to_dict()} (NaN: {df[name].isnull().sum()})")
```

```python
# Verify Playground data relationships
print("VERIFICATION - Playgrounds:")
print("=" * 80)

# Check Male playgrounds
male_none_new = 'Male None Playground'
male_sport_new = [name for _, name in male_playgrounds if 'None' not in name]

print("\nMALE PLAYGROUNDS:")
print(f"Rows where Male None = 1: {(df[male_none_new] == 1).sum()}")
print("  Checking if other male sports are 0 when None=1...")
mask_male_none = df[male_none_new] == 1
for sport in male_sport_new:
    has_sport = (df.loc[mask_male_none, sport] == 1).sum()
    if has_sport > 0:
        print(f"    WARNING: {sport} has {has_sport} entries = 1 when None=1!")
    else:
        print(f"    ✓ {sport}: All 0 when None=1")

print(f"\nRows where at least one male sport = 1: {(df[male_sport_new] == 1).any(axis=1).sum()}")
print(f"Rows where all male playgrounds are NaN: {df[male_sport_new + [male_none_new]].isnull().all(axis=1).sum()}")

# Check Female playgrounds
female_none_new = 'Female None Playground'
female_sport_new = [name for _, name in female_playgrounds if 'None' not in name]

print("\n\nFEMALE PLAYGROUNDS:")
print(f"Rows where Female None = 1: {(df[female_none_new] == 1).sum()}")
print("  Checking if other female sports are 0 when None=1...")
mask_female_none = df[female_none_new] == 1
for sport in female_sport_new:
    has_sport = (df.loc[mask_female_none, sport] == 1).sum()
    if has_sport > 0:
        print(f"    WARNING: {sport} has {has_sport} entries = 1 when None=1!")
    else:
        print(f"    ✓ {sport}: All 0 when None=1")

print(f"\nRows where at least one female sport = 1: {(df[female_sport_new] == 1).any(axis=1).sum()}")
print(f"Rows where all female playgrounds are NaN: {df[female_sport_new + [female_none_new]].isnull().all(axis=1).sum()}")

# Overall distribution
print("\n\nOVERALL DISTRIBUTION:")
for _, name in all_playgrounds:
    count_1 = (df[name] == 1).sum()
    count_0 = (df[name] == 0).sum()
    count_nan = df[name].isnull().sum()
    print(f"{name}: Yes={count_1}, No={count_0}, NaN={count_nan}")
```

4. Number of Farms in Mouza Premises 4-Fish Farms

**Cleaning Strategy:**
- Mutually exclusive logic:
  - If "None" column has value 1 → All other farm columns = 0
  - If any other farm column has a count value → "None" column = 0
  - If ALL 4 columns are empty → NaN for all
- **Renamed columns:**
  - "No Farms" (binary: 1=no farms, 0=has farms)
  - "Number of Poultry Farms"
  - "Number of Livestock/Dairy Farms"
  - "Number of Fish Farms"

```python
# Inspect Farm columns before cleaning
farm_cols = {
    'none': 'Number of Farms in Mouza Premises \n1-None',
    'poultry': 'Number of Farms in Mouza Premises \n2-ultry Farms',
    'livestock': 'Number of Farms in Mouza Premises \n3-Livestock/Dairy Farms',
    'fish': 'Number of Farms in Mouza Premises \n4-Fish Farms'
}

print("BEFORE CLEANING - Farm Types:")
print("=" * 80)

for farm_type, col in farm_cols.items():
    print(f"\n{farm_type.upper()}:")
    print(f"  Missing: {df[col].isnull().sum()}")
    unique_vals = df[col].dropna().unique()
    print(f"  Unique values (first 10): {list(unique_vals)[:10]}")
    print(f"  Value counts: {df[col].value_counts().head(10).to_dict()}")
    print(f"  Sample values: {df[col].dropna().head(5).tolist()}")
```

```python
# Clean Farm columns with mutually exclusive logic
print("CLEANING Farm Types...")
print("=" * 80)

# Step 1: Create mask for ALL empty rows (all 4 farm columns are empty)
all_farm_cols = list(farm_cols.values())
mask_all_empty = df[all_farm_cols].isnull().all(axis=1) | \
                 (df[all_farm_cols].astype(str).apply(lambda x: x.str.strip()).replace('', np.nan).isnull().all(axis=1))

print(f"\nRows where ALL farm columns are empty: {mask_all_empty.sum()}")

# Step 2: Clean "None" column (convert to binary)
none_col = farm_cols['none']
df[none_col] = pd.to_numeric(df[none_col], errors='coerce')
# Convert to binary: 1 stays 1, anything else becomes 0
df[none_col] = df[none_col].apply(lambda x: 1 if x == 1 else (0 if pd.notna(x) else np.nan))

# Step 3: Clean numeric farm columns (Poultry, Livestock, Fish)
for farm_type in ['poultry', 'livestock', 'fish']:
    col = farm_cols[farm_type]
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(0).astype(int)

# Step 4: Apply mutually exclusive logic
# If None=1, then all other farms must be 0
mask_none = df[none_col] == 1
for farm_type in ['poultry', 'livestock', 'fish']:
    df.loc[mask_none, farm_cols[farm_type]] = 0
print(f"Rows where None=1 (no farms): {mask_none.sum()}")

# If any farm has a value > 0, then None must be 0
mask_has_farms = (df[farm_cols['poultry']] > 0) | \
                 (df[farm_cols['livestock']] > 0) | \
                 (df[farm_cols['fish']] > 0)
df.loc[mask_has_farms, none_col] = 0
print(f"Rows with at least one farm type: {mask_has_farms.sum()}")

# Step 5: Restore NaN for rows where ALL columns were originally empty
for col in all_farm_cols:
    df.loc[mask_all_empty, col] = np.nan

print(f"Rows restored to NaN (no farm data): {mask_all_empty.sum()}")

# Step 6: Rename columns
rename_dict = {
    farm_cols['none']: 'No Farms',
    farm_cols['poultry']: 'Number of Poultry Farms',
    farm_cols['livestock']: 'Number of Livestock/Dairy Farms',
    farm_cols['fish']: 'Number of Fish Farms'
}
df.rename(columns=rename_dict, inplace=True)

print("\nAFTER CLEANING - Farm Types:")
print("=" * 80)
print(f"\nNo Farms (binary):")
print(f"  Value counts: {df['No Farms'].value_counts().to_dict()}")
print(f"  NaN count: {df['No Farms'].isnull().sum()}")

for farm_name in ['Number of Poultry Farms', 'Number of Livestock/Dairy Farms', 'Number of Fish Farms']:
    print(f"\n{farm_name}:")
    print(f"  Min: {df[farm_name].min()}, Max: {df[farm_name].max()}, Mean: {df[farm_name].mean():.2f}")
    print(f"  NaN count: {df[farm_name].isnull().sum()}")
    print(f"  Zero count: {(df[farm_name] == 0).sum()}")
    print(f"  Positive count: {(df[farm_name] > 0).sum()}")
```

## Section 38: Natural Disasters

Clean natural disaster data:
- Main column: 1 = Yes (disaster occurred), 2→0 = No, 0/NaN → NaN
- Disaster types (Drought, Flood, Earthquake, Others): 
  - If disaster=Yes: keep marked types, fill unmarked with 0
  - If disaster=No: all types = 0
  - If disaster=NaN: all types = NaN

```python
# Clean Natural Disaster Columns (Columns 173-177)

disaster_col_idx = 173  # Main yes/no column
disaster_type_cols_idx = [174, 175, 176, 177]  # Drought, Flood, Earthquake, Others

print("Cleaning natural disaster columns...")

# Get column names
main_col = df.columns[disaster_col_idx]
type_cols = [df.columns[i] for i in disaster_type_cols_idx]

# Clean main disaster column
main_series = df.iloc[:, disaster_col_idx].copy()
main_series = main_series.apply(lambda x: np.nan if isinstance(x, str) and (x.startswith('P') or x.startswith('N')) else x)
main_series = pd.to_numeric(main_series, errors='coerce')

# Create masks
mask_yes = main_series == 1  # Disaster occurred
mask_no = main_series == 2  # No disaster
mask_no_data = (main_series == 0) | main_series.isna()  # No data

# Convert main column: 1→1, 2→0, 0/NaN→NaN
main_new = main_series.copy()
main_new[mask_yes] = 1
main_new[mask_no] = 0
main_new[mask_no_data] = np.nan

# Clean disaster type columns
type_series_list = []
for i, col in zip(disaster_type_cols_idx, type_cols):
    series = df.iloc[:, i].copy()
    series = series.apply(lambda x: np.nan if isinstance(x, str) and (x.startswith('P') or x.startswith('N')) else x)
    series = pd.to_numeric(series, errors='coerce')
    
    series_new = series.copy()
    # When disaster occurred: if type not marked, set to 0
    series_new[mask_yes] = series_new[mask_yes].fillna(0)
    # When no disaster: set all types to 0
    series_new[mask_no] = 0
    # When no data: set all types to NaN
    series_new[mask_no_data] = np.nan
    
    type_series_list.append(series_new)

# New column names
new_col_names = [
    'Natural Disaster in Last 5 Years',
    'Disaster Type: Drought',
    'Disaster Type: Flood',
    'Disaster Type: Earthquake',
    'Disaster Type: Others'
]

# Replace columns
old_cols_to_drop = [main_col] + type_cols
df = df.drop(columns=old_cols_to_drop)

df.insert(disaster_col_idx, new_col_names[0], main_new)
for i, (new_name, new_series) in enumerate(zip(new_col_names[1:], type_series_list), 1):
    df.insert(disaster_col_idx + i, new_name, new_series)

print("✅ Natural disaster columns cleaned!")
```

## Section 47: PCO and Mobile Signals Facility Pairs

Columns 158-161: 2 facility pairs (PCO and Mobile Signals)

**Pattern:**
- **Availability column** originally had distance values mixed with availability
- **Logic:**
  - If original value = 0 → Facility available in mouza (convert to 1, distance = 0)
  - If original value > 0 → Not available, that value is the distance (convert to 0, distance = value)
  - If both empty → Keep as NaN

**Final encoding:**
- Availability: 1 = In mouza, 0 = Not in mouza, NaN = No data
- Distance: 0 = In mouza, >0 = Distance in km, NaN = No data

```python
# Clean PCO and Mobile Signals Facility Pairs (Columns 158-161)

print("Cleaning PCO and Mobile Signals facility pairs...")
print("="*80)

# Define the pairs
pairs = [
    ('PCO Available', 'PCO Distance (km)'),
    ('Mobile Signals Available', 'Mobile Signals Distance (km)')
]

for avail_col, dist_col in pairs:
    print(f"\nProcessing: {avail_col} / {dist_col}")
    print("-" * 80)
    
    avail_series = df[avail_col].copy()
    dist_series = df[dist_col].copy()
    
    # Check for both empty (NaN in both)
    mask_both_empty = avail_series.isna() & dist_series.isna()
    print(f"  Rows with both empty: {mask_both_empty.sum()}")
    
    # Create new cleaned series
    new_avail = pd.Series(index=df.index, dtype='float64')
    new_dist = pd.Series(index=df.index, dtype='float64')
    
    # Case 1: Both empty -> keep as NaN
    new_avail[mask_both_empty] = np.nan
    new_dist[mask_both_empty] = np.nan
    
    # Case 2: Availability = 0 -> Available in mouza (avail=1, dist=0)
    mask_available = (avail_series == 0) & ~mask_both_empty
    new_avail[mask_available] = 1
    new_dist[mask_available] = 0
    print(f"  Rows where available in mouza (0→1, dist=0): {mask_available.sum()}")
    
    # Case 3: Availability > 0 -> Not available, value is distance (avail=0, dist=value)
    mask_not_available = (avail_series > 0) & ~mask_both_empty
    new_avail[mask_not_available] = 0
    new_dist[mask_not_available] = avail_series[mask_not_available]
    print(f"  Rows where not available (>0→0, dist=value): {mask_not_available.sum()}")
    
    # Case 4: Handle cases where distance column has value but availability is NaN
    mask_dist_only = avail_series.isna() & dist_series.notna()
    if mask_dist_only.sum() > 0:
        # If distance = 0, means available
        mask_dist_zero = mask_dist_only & (dist_series == 0)
        new_avail[mask_dist_zero] = 1
        new_dist[mask_dist_zero] = 0
        
        # If distance > 0, means not available
        mask_dist_nonzero = mask_dist_only & (dist_series > 0)
        new_avail[mask_dist_nonzero] = 0
        new_dist[mask_dist_nonzero] = dist_series[mask_dist_nonzero]
        
        print(f"  Rows with only distance data: {mask_dist_only.sum()}")
    
    # Update dataframe
    df[avail_col] = new_avail
    df[dist_col] = new_dist
    
    # Verify
    print(f"\n  After cleaning:")
    print(f"    Available: 1={new_avail[new_avail==1].count()}, 0={new_avail[new_avail==0].count()}, NaN={new_avail.isna().sum()}")

print("\n✅ PCO and Mobile Signals pairs cleaned!")

# Verification
print("\n" + "="*80)
print("CONSISTENCY CHECK:")
print("="*80)

for avail_col, dist_col in pairs:
    violations_1 = ((df[avail_col] == 1) & (df[dist_col] != 0)).sum()
    violations_2 = ((df[avail_col] == 0) & (df[dist_col] == 0)).sum()
    
    print(f"\n{avail_col}:")
    print(f"  Violations (avail=1 but dist≠0): {violations_1}")
    print(f"  Violations (avail=0 but dist=0): {violations_2}")
```

1. If Available = 1.0 → facility is available, so Distance should be 0
2. If both Available and Distance are empty (NaN) → keep both as NaN  
3. Any value other than 1 in Available (like 0, 2, etc.) → should become 0

**Pairs to clean:**
- PCO Available / PCO Distance (km)
- Mobile Signals Available / Mobile Signals Distance (km)
- Internet: Fixed Line DSL Available / Distance
- Internet: Cable Net Available / Distance
- Internet: Mobile Net Available / Distance

```python
# Define the facility pairs to clean
pco_mobile_internet_pairs = [
    ('PCO Available', 'PCO Distance (km)'),
    ('Mobile Signals Available', 'Mobile Signals Distance (km)'),
    ('Internet: Fixed Line DSL Available', 'Internet: Fixed Line DSL Distance (km)'),
    ('Internet: Cable Net Available', 'Internet: Cable Net Distance (km)'),
    ('Internet: Mobile Net Available', 'Internet: Mobile Net Distance (km)')
]

# Check current state before cleaning
print("=== Before Cleaning ===\n")
for avail_col, dist_col in pco_mobile_internet_pairs:
    avail_unique = df[avail_col].dropna().unique()
    dist_null = df[dist_col].isna().sum()
    avail_null = df[avail_col].isna().sum()
    both_null = ((df[avail_col].isna()) & (df[dist_col].isna())).sum()
    
    print(f"{avail_col}:")
    print(f"  Unique values: {sorted(avail_unique)}")
    print(f"  NaN count: {avail_null}")
    print(f"{dist_col}:")
    print(f"  NaN count: {dist_null}")
    print(f"  Both NaN: {both_null}")
    print()
```

```python
# Apply cleaning rules
print("=== Applying Cleaning Rules ===\n")

for avail_col, dist_col in pco_mobile_internet_pairs:
    # Count changes
    changes_made = 0
    
    # Rule 1: If Available = 1.0, set Distance to 0
    mask_avail_1 = (df[avail_col] == 1.0)
    if mask_avail_1.any():
        df.loc[mask_avail_1, dist_col] = 0.0
        changes_made += mask_avail_1.sum()
        print(f"{avail_col}: Set distance to 0 for {mask_avail_1.sum()} rows where Available=1")
    
    # Rule 2: Both NaN → keep as NaN (no action needed)
    both_nan = ((df[avail_col].isna()) & (df[dist_col].isna())).sum()
    print(f"{avail_col}: {both_nan} rows have both columns as NaN (kept as-is)")
    
    # Rule 3: Any value other than 1 in Available → convert to 0
    mask_not_1 = (df[avail_col].notna()) & (df[avail_col] != 1.0)
    if mask_not_1.any():
        df.loc[mask_not_1, avail_col] = 0.0
        print(f"{avail_col}: Converted {mask_not_1.sum()} non-1 values to 0")
    
    print()

print("✅ All PCO, Mobile Signals, and Internet facility pairs cleaned!")
```

```python
# Verify the cleaning
print("=== After Cleaning ===\n")
for avail_col, dist_col in pco_mobile_internet_pairs:
    avail_unique = df[avail_col].dropna().unique()
    dist_null = df[dist_col].isna().sum()
    avail_null = df[avail_col].isna().sum()
    both_null = ((df[avail_col].isna()) & (df[dist_col].isna())).sum()
    
    # Check for any violations
    violation_1_not_0 = ((df[avail_col] == 1.0) & (df[dist_col] != 0.0)).sum()
    
    print(f"{avail_col}:")
    print(f"  Unique values: {sorted(avail_unique)}")
    print(f"  NaN count: {avail_null}")
    print(f"{dist_col}:")
    print(f"  NaN count: {dist_null}")
    print(f"  Both NaN: {both_null}")
    print(f"  ⚠️ Violations (Available=1 but Distance≠0): {violation_1_not_0}")
    print()
```

```python
# Save cleaned dataset
df.to_csv('Mouza Census-2020-Cleaned.csv', index=False)
print("\n✅ Dataset saved with PCO, Mobile Signals, and Internet facilities cleaned!")
```
